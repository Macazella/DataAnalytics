📦 1. Carga del Dataset y Librerías  
Carga del dataset final y librerías de análisis.


```python
# ✅ Cargar librerías necesarias y dataset
import pandas as pd

df = pd.read_csv("RUTA/AL/DATASET_FINAL.csv")  # <- cambiar ruta
```

🎯 2. Preparación de KPIs para Reporte  
Se generan indicadores clave para visualización.


```python
# ✅ Cálculo de KPIs generales
kpi_attrition = df['Attrition'].value_counts(normalize=True) * 100
prom_ingreso = df['MonthlyIncome'].mean()
print(f"Tasa de Attrition:\n{kpi_attrition}")
print(f"Ingreso promedio: ${prom_ingreso:.2f}")
```

📊 3. Tablas Dinámicas para Visualización  
Se construyen tablas resumen por dimensiones clave.


```python
# ✅ Pivot tables
pivot_dep = df.pivot_table(index='Department', values='MonthlyIncome', aggfunc='mean')
pivot_rol = df.pivot_table(index='JobRole', values='Attrition', aggfunc=lambda x: (x=='Yes').mean())

display(pivot_dep)
display(pivot_rol)
```

📤 4. Exportación Final para BI  
Se exporta el archivo final con columnas clave y KPIs listos.


```python
# ✅ Guardar dataset reducido
columnas_bi = ['EmployeeID', 'Attrition', 'JobRole', 'Department', 'MonthlyIncome', 'YearsAtCompany']
df[columnas_bi].to_csv("HR_Analytics_para_PowerBI.csv", index=False)
```

✅ 5. Conclusión Final  
Se valida que el archivo exportado contenga los datos necesarios.


```python
# ✅ Validación y cierre del proyecto
df_bi = pd.read_csv("HR_Analytics_para_PowerBI.csv")
assert set(columnas_bi).issubset(df_bi.columns)
print("✅ Exportación final realizada correctamente.")
print("🎯 El proyecto está listo para presentarse en Power BI.")
```
