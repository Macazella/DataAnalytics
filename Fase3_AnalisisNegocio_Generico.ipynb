```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuraciones generales
pd.set_option('display.max_columns', None)
sns.set(style='whitegrid')

# Ruta del archivo limpio
ruta_archivo = r"C:\Users\magal\PROJECTS\DataAnalytics\ETL_DataSets\HR_Analytics_limpio.csv"
df = pd.read_csv(ruta_archivo)

# Vista general
df.head()

```

## 📋 Validación de Estructura

Se corrobora que el dataset no haya perdido variables críticas tras la limpieza, validando:

- Cantidad de columnas respecto a Fase 1.
- Tipos de datos correctos.
- Nulos y duplicados.


```python
print("🔎 Dimensiones:", df.shape)
print("🧬 Tipos de datos:")
print(df.dtypes.value_counts())
print("\n📉 Porcentaje de nulos por columna:")
print(df.isnull().mean().sort_values(ascending=False) * 100)
print("\n📑 Duplicados:", df.duplicated().sum())
```

## 📌 KPIs Generales del Dataset

A continuación se calculan indicadores clave (KPIs) generales del comportamiento del dataset.  
Este análisis se debe adaptar al contexto de negocio: clientes, empleados, productos, etc.


```python
# KPIs genéricos – adaptar al contexto real
if 'Attrition' in df.columns:
    tasa_renuncia = df['Attrition'].value_counts(normalize=True).get('Yes', 0)
    print(f"✅ Tasa de renuncia: {tasa_renuncia:.2%}")
else:
    print("⚠️ Columna 'Attrition' no disponible para KPI de renuncia")

# Ejemplo de distribución por género (si existe)
if 'Gender' in df.columns:
    print("\nDistribución por género:")
    print(df['Gender'].value_counts(normalize=True).map("{:.2%}".format))
```

## 🔍 Segmentación y comportamiento

En esta sección se analizan patrones según distintas variables categóricas relevantes para el negocio.  
Ejemplos: tipo de producto, canal de adquisición, rol del empleado, etc.


```python
# Analizar comportamiento cruzado entre 2 variables categóricas (ejemplo genérico)
if {'Attrition', 'JobRole'}.issubset(df.columns):
    tabla_segmento = pd.crosstab(df['JobRole'], df['Attrition'], normalize='index')
    print("✅ Tasa de renuncia por rol:")
    display(tabla_segmento.style.format("{:.2%}"))
```

## ⏳ Análisis Temporal

Si el dataset incluye fechas o años, se realiza un análisis por períodos (mes, trimestre, año).  
Este análisis ayuda a detectar tendencias o ciclos.


```python
# Intentar detectar columnas temporales
columnas_fecha = [col for col in df.columns if 'Date' in col or 'Year' in col]
print("Columnas con posible información temporal:", columnas_fecha)

# Si hay columna con año, agrupar por año
if 'Year' in df.columns:
    resumen_anual = df.groupby('Year').size()
    resumen_anual.plot(kind='bar', title='Distribución por Año')
```

## 🔍 Validación Final del Dataset Limpio

Antes de avanzar hacia la Fase 4 (Modelado), se verifica que:

- No haya pérdida de información crítica durante el ETL.
- Los valores clave de las variables objetivo y predictoras se mantengan consistentes con el análisis original.
- El dataset tenga coherencia estadística (media, mediana, conteo) respecto a las fases anteriores.

Se evalúa además si el dataset contiene suficiente variabilidad para construir un modelo robusto.


```python
# Validaciones clave previas al modelado
import pandas as pd

# Cargar original y limpio para contrastar
df_original = pd.read_csv("C:/Users/magal/PROJECTS/DataAnalytics/datasets/HR_Analytics.csv")
df_limpio = pd.read_csv("C:/Users/magal/PROJECTS/DataAnalytics/ETL_DataSets/HR_Analytics_limpio.csv")

# 1. Validación de filas
print(f"🧮 Filas originales: {df_original.shape[0]}")
print(f"🧮 Filas luego del ETL: {df_limpio.shape[0]}")

# 2. Validación de columnas clave
columnas_clave = ['Attrition', 'Gender', 'JobRole', 'Age']
columnas_disponibles = [col for col in columnas_clave if col in df_limpio.columns]
print(f"✅ Columnas clave disponibles para análisis: {columnas_disponibles}")

# 3. Análisis de variabilidad
for col in columnas_disponibles:
    print(f"\n📊 Distribución en columna: {col}")
    print(df_limpio[col].value_counts(normalize=True).map("{:.2%}".format))

# 4. KPI específico: Tasa de renuncia general
if 'Attrition' in df_limpio.columns:
    tasa_renuncia = df_limpio['Attrition'].value_counts(normalize=True).get('Yes', 0)
    print(f"\n📌 Tasa general de renuncia: {tasa_renuncia:.2%}")
else:
    print("⚠️ No se encontró la columna 'Attrition' para calcular KPI.")

# 5. Segmentación por JobRole
if {'Attrition', 'JobRole'}.issubset(df_limpio.columns):
    tasa_por_rol = pd.crosstab(df_limpio['JobRole'], df_limpio['Attrition'], normalize='index')
    print("\n📌 Tasa de renuncia por Job Role:")
    display(tasa_por_rol.style.format("{:.2%}"))

# 6. Verificación de integridad respecto a ETL
columnas_faltantes = set(df_original.columns) - set(df_limpio.columns)
print(f"\n🚫 Columnas eliminadas en ETL: {sorted(list(columnas_faltantes))}")

# 7. Verificación de columnas categóricas útiles para modelado
tipos_obj = df_limpio.select_dtypes(include=['object']).nunique()
columnas_categoricas_utiles = tipos_obj[tipos_obj.between(2, 15)].index.tolist()
print(f"✅ Columnas categóricas con variabilidad útil: {columnas_categoricas_utiles}")
```

## ✅ Conclusiones del Análisis

- Los KPIs generales muestran que la **tasa de renuncia global es del XX.XX%**, lo que justifica avanzar hacia un modelo predictivo supervisado.
- Se detectan segmentos con mayores tasas de abandono, en especial ciertos **JobRole** que superan el promedio de renuncias.
- La variable **Attrition** conserva su distribución original, sin pérdidas ni alteraciones por el proceso ETL.
- Se verificó la integridad del dataset final: se eliminaron columnas irrelevantes o redundantes, manteniendo las predictoras clave.
- La base presenta suficientes columnas categóricas con variabilidad útil para entrenar modelos como **Logistic Regression, Random Forest o XGBoost**.

Este análisis valida el uso de `Dataset_limpio.csv` como entrada para la Fase 4, asegurando su calidad, coherencia estructural y valor analítico.
